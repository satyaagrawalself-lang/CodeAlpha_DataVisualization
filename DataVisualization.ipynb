```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("books_data.csv")
```

Step 2: Price Distribution Chart

```python
plt.figure(figsize=(8,5))
plt.hist(df["Price"], bins=5)
plt.title("Distribution of Book Prices")
plt.xlabel("Price")
plt.ylabel("Number of Books")
plt.show()
```

Step 3: Rating Distribution Chart

```python
rating_count = df["Rating"].value_counts()

plt.figure(figsize=(8,5))
rating_count.plot(kind="bar")
plt.title("Book Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.show()
```

Step 4: Top 10 Most Expensive Books

```python
top_books = df.sort_values("Price", ascending=False).head(10)

plt.figure(figsize=(10,5))
plt.bar(top_books["Title"], top_books["Price"])
plt.title("Top 10 Most Expensive Books")
plt.xlabel("Book Title")
plt.ylabel("Price")
plt.xticks(rotation=90)
plt.show()
```

Step 5: Pie Chart of Ratings

```python
plt.figure(figsize=(7,7))
df["Rating"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)
plt.title("Percentage Distribution of Ratings")
plt.ylabel("")
plt.show()
```

Key Findings

• Book prices are distributed across different price ranges.

• One-star ratings are the most common category.

• Some books are significantly more expensive than others.

• Rating distribution shows a mix of customer preferences.

• Visualizations help identify patterns and trends in the dataset.
